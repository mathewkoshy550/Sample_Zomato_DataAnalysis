```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
data = pd.read_csv("/content/fraud_transactions.csv")
data.head(5)
```

```python
x=data.drop(['is_fraud'], axis=1)
y=data['is_fraud']
#check what it does
```

```python
data.info()
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.85)
#85% used for training
```

```python
x_train.shape[1]
 # these 10 nuerons will be used for building the first layer of the nueral network
```

```python
model = keras.Sequential([
layers.Input(shape=(x_train.shape[1],),name='input'),
layers.Dense(16, activation='relu',name='hidden_layer_1'),
layers.Dense(32, activation='relu',name='hidden_layer_2'),
layers.Dense(1, activation='relu',name='output_layer')])
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(x_train,y_train,epochs=10)
```
